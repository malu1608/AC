```python
players_teams_data = pd.read_csv("../data/players_teams.csv")
players_teams_data.head()

```

```python
players_data = pd.read_csv("../data/players.csv")
print(players_data.head())

valores_omissos = players_data.isnull().sum()

# Calcule a percentagem de omissão em relação ao número total de linhas
percentagem_omissao = (valores_omissos / len(players_data)) * 100

# Crie um novo DataFrame para exibir os resultados
resultados = pd.DataFrame({'Valores Omissos': valores_omissos, 'Percentagem de Omissão': percentagem_omissao})

print("\n")
resultados_filtrados = resultados[resultados['Percentagem de Omissão'] != 0]
print(resultados_filtrados)
```
