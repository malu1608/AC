```python
import pandas as pd

players_teams_data = pd.read_csv("../data/players_teams.csv")
```

```python
# delete players with no attempts
players_teams_data = players_teams_data[players_teams_data['fgAttempted'] != 0]

players_teams_data['ORTG'] = (players_teams_data['points'] / players_teams_data['fgAttempted']) * 100

""" players_teams_data['DRTG'] = (players_teams_data['points'] / players_teams_data['fgAttempted']) * 100 """

""" players_teams_data['UE'] = (players_teams_data['fgAttempted'] + 0.44 * players_teams_data['ftAttempted'] + players_teams_data['turnovers']) * 100 / players_teams_data['minutes'] """

```

```python
players_teams_data['ORTG'] = players_teams_data['ORTG'].apply(lambda x: round(x, 1))

# Display the players team table with ORTG
print(players_teams_data[['playerID', 'year', 'ORTG']])
```

```python
players_teams_data.drop(["points", "fgAttempted"], axis=1, inplace=True)
players_teams_data
```

```python
# Verificar valores ausentes em todo o DataFrame
missing_rows = players_teams_data[players_teams_data.isnull().any(axis=1)]

colunas = ['playerID', 'fgAttempted', 'ftAttempted', 'turnovers', 'minutes']

# Imprimir as linhas com valores ausentes
print("Linhas com missing values:")
print(missing_rows[colunas])
```
