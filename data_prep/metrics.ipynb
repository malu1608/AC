```python
import pandas as pd

players_teams_data = pd.read_csv("../data/players_teams.csv")
```

```python
players_teams_data['ORTG'] = (players_teams_data['points'] / players_teams_data['fgAttempted']) * 100

players_teams_data['DRTG'] = (players_teams_data['points'] / players_teams_data['fgAttempted']) * 100

players_teams_data['UE'] = (players_teams_data['fgAttempted'] + 0.44 * players_teams_data['ftAttempted'] + players_teams_data['turnovers']) * 100 / players_teams_data['minutes']

# Display the players team table with ORTG, DRTG and UE
print(players_teams_data[['playerID', 'year', 'ORTG', 'DRTG', 'UE']])
```
