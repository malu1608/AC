```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

# Diretório do arquivo CSV
dir = r"data\\teams.csv"

# Carrega os dados do CSV
df = pd.read_csv(dir)

colunas_para_remover = ['seeded', 'lgID','divID']
df = df.drop(columns=colunas_para_remover)

# Colunas a serem codificadas
columns_to_encode = ['franchID', 'confID', 'name', 'tmID']

df_encoded = pd.get_dummies(df, columns=columns_to_encode)

# Defina a variável-alvo (target)
target = 'playoff'

# Garanta que a coluna 'playoff' seja do tipo inteiro
df_encoded['playoff'] = df_encoded['playoff'].apply(lambda x: 1 if x == 'Y' else 0)

# Especifique a lista de características (features) que você deseja usar para treinamento
features = df_encoded.columns.difference([target])  # Todas as colunas, exceto 'playoff'

df_encoded['arena'], arena_mapping = pd.factorize(df['arena'])
df_encoded['firstRound'] = df_encoded['firstRound'].replace({'L': 0, 'W': 1})
df_encoded['semis'] = df_encoded['semis'].replace({'L': 0, 'W': 1})
df_encoded['finals'] = df_encoded['finals'].replace({'L': 0, 'W': 1})
df_encoded = df_encoded.dropna(subset=['firstRound', 'semis', 'finals'])

# Divide os dados em treinamento e teste usando o DataFrame codificado
X_train, X_test, y_train, y_test = train_test_split(df_encoded[features], df_encoded[target], test_size=0.3, random_state=42)

# Cria o modelo de Árvore de Decisão
dt = DecisionTreeClassifier()

# Treina o modelo com os dados de treinamento
dt.fit(X_train, y_train)

# Faz previsões com os dados de teste
y_pred = dt.predict(X_test)

# Avalia a precisão do modelo
accuracy = accuracy_score(y_test, y_pred)
print("A precisão do modelo é:", accuracy)
```
